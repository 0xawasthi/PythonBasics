````python
# variable length argument in python

def add(a, b, c, d):
    return a+b+c+d

sum = add(1, 2, 3)
print(sum)
````

In the above example, the number of arguments is fixed, thus while calling the function, if no. of argument is not matched. Then in this case you'll get an error.

To avoid this issue:

you can use variable length argument

- positional arguments and keyword arguments

````python
def print_numbers(*args):
    for numbers in args:
        print(numbers)
````

````python
print_numbers(1, 2, 3, 4, 5, "ElevateX")
````

In the above example, * means to collect positional arguments whereas args is the variable name.

````python
# Similarly lets see the concept of **kwargs

def details(**info):
    print(info)

details(name = "Aashish", org = "Elevate-X")
````

````python
# Example 2: Password Strength Checker
# Condition for strong password
#   1. length should 8 or above
#   2. password should contain at least 1 digit
#   3. at least one letter should be uppercase and one should be lowercase
#   4. should use special character like !@#$%^


def check_password(password):
    if len(password) >= 8 and any(letter.isdigit() for letter in password) and any(letter.isupper() for letter in password) and any(letter.islower() for letter in password) and any(letter in "!@#$$%^&*" for letter in password):
        print("Strong Password.")
    else:
        print("Weak Password.")
````

````python
check_password("Aashish@123")
````

````python
check_password("aashish0@")
````

# Lamda function



Lamda Function in Python:

Lamda functions are small anonymous function defined using the keyword ```lamda```. It can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher order function.

````python
from ast import Expression

## Syntax

lambda arguments : Expression
````

````python
def add(a,b):
  return(a+b)

add(2, 3)
````

````python
add = lambda a, b: a+b
print(add(2,3))
````

````python
# Check whether the number is even or odd using lambda function:

even_odd = lambda num: "even" if num % 2 == 0 else "odd"
print(even_odd(2))
print(even_odd(3))
````

# map() function in python

````python
def square(x):
    return x*x
````

````python
square(10)
````

````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
list(map(square, numbers))
````

````python
# lambda function with map

list(map(lambda x: x*x, numbers))
````

````python
# use map to convert a list of strings to integers

str_num = ['1', '2', '3']

int_num = list(map(int, str_num))

int_num
````

````python
# Convert each word to Uppercase

words = ["apple", "ball", "cat"]

list(map(str.upper, words))
````
