```python
#@ Read a whole file
with open('elevatex.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
#@ Read a file line by line
with open('elevatex.txt', 'r') as file:
    content = file.readlines()
    print(content)
```

```python
#@ Writing a file (Overwriting)
with open('elevatex.txt', 'w') as file:
    file.write('Hello World!\n')
    file.write('I am Aashish Awasthi \n')
    file.write('I am from Elevate-X')
```

```python
#@ Read a whole file
with open('elevatex.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
#@ What if I dont want to remove the existing content, whereas i am concerned about adding a new lines in a file
#@ without overwriting
#@ we use the mode append 'a'
with open('elevatex.txt', 'a') as file:
    file.write('\nWe are from Bharatpur, Chitwan')
```

```python
#@ Read a whole file
with open('elevatex.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
#@ What if I need to copy a text file source to a destination text file
# copying a file to a new file
with open('elevatex.txt', 'r') as source_file:
    content = source_file.read()
with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)
```

```python
#@ Read a whole file
with open('destination.txt', 'r') as file:
    content = file.read()
    print(content)
```

# Exception Handling

- Understanding Exceptions
  - Run time error handling
  - If some run time, logical error aayo vane pani, it makes sure that the corrective actions will be taken without stopping the execution of the program
  

Exception:
- Exception can be defined as an event that disturbs normal flow of a program.
- It occurs when an error is encountered during the program execution.

Common exceptions includes:
- ZeroDivisionError: Divide by zero
- FileNotFoundError: File not found
- ValueError: Invalid value
- TypeError: Invalid type

To handle exception, we use:
- try
- except
- else
- finally

```python
# Exception try, except block

try:
  # Statement
  a = b
except:
  # Statement
  print("The variable has not been defined.")
```

```python
a = b
print(a)
```

For the above example, we gave the error message by ourself,
We can directly print the error message as well.

When the backend api gives an error, the frontend will show the same error comming from the backend api

```python
try:
  a = b
except NameError as e:
  print(e)
```

```python
# Let's implement ZeroDivisionError

result = 5/0
result
```

```python
try:
  result = 5/0
except ZeroDivisionError as er:
  print(er)
  print("Denominator should always be greater than 0.")
```

```python
try:
  num = int(input("Enter a number: "))
  result = 10/num
  print("No Exception")
except ZeroDivisionError as ex:
  print(ex)
  print("Denominator should always be greater than 0.")
except NameError as ex1:
  print(ex1)
  print("NameError occured.")
except ValueError as ex2:
  print(ex2)
  print("Please enter a valid number.")
except Exception as ex3:
  print(ex3)
  print("Something went wrong.")
```

```python
#@ try, except and else block

try:
  num = int(input("Enter a number: "))
  result = 10/num
except ZeroDivisionError as ex:
  print(ex)
  print("Denominator should always be greater than 0")
except NameError as ex1:
  print(ex1)
  print("<<<<<<<<<<<<<<<<Main exception is caught here>>>>>>>>>>>>>>")
except Exception as ex2:
  print(ex2)
  print("Something went wrong.")
else:
  print("No exception occured")
  print(result)
```

```python
#@ try, except, else and finally block

try:
  num = int(input("Enter a number: "))
  result = 10/num
except ZeroDivisionError as ex:
  print(ex)
  print("Denominator should always be greater than 0")
except NameError as ex1:
  print(ex1)
  print("<<<<<<<<<<<<<<<<Main exception is caught here>>>>>>>>>>>>>>")
except Exception as ex2:
  print(ex2)
  print("Something went wrong.")
else:
  print("No exception occured")
  print(result)
finally:
  print("Execution is completed.")
```
