# Multiple Methods Examples

```python
class BankAccount:
    def __init__(self, owner):
        self.owner = owner
        self.balance = 0.00

    def deposit(self, amount):
        self.balance += amount
        print(f"{amount} deposited to {self.owner}'s account.")

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            print(f"{amount} withdrawn from {self.owner}'s account.")
        else:
            print(f"Insufficient balance in {self.owner}'s account.")

    def display_statement(self):
        print(f"Account balance for {self.owner}: ${self.balance}")

```
