# Multiple Methods Examples

***1. The Bank Account Tracker***

**Goal:** Build a basic account management class where methods depend directly on the values altered by other methods.
- **Class Name:** BankAccount
- **The Methods:**
  - __init__(self, owner, balance=0.0): Initializes the account owner's name and starting balance.
  - deposit(self, amount): Adds money to the balance and automatically prints the new total.
  - withdraw(self, amount): Checks if there is enough money. If yes, deducts it and prints the amount withdrawn. If not, prints "Insufficient funds".
  - display_statement(self): Prints a summary of the current owner and balance.
- **How They Work Together:** Your withdraw method must read the balance updated by the deposit method before deciding whether to allow the transaction.

```python
class BankAccount:
    def __init__(self, owner):
        self.owner = owner
        self.balance = 0.00

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Rs. {amount} deposited to {self.owner}'s account.")
            print(f"Available balance for {self.owner}'s account is Rs. {self.balance} ")
        else:
            print("Invalid deposit amount. Amount must be greater than zero.")
        print()

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount. Amount must be greater than zero.")
            return
        if self.balance >= amount:
            self.balance -= amount
            print(f"Rs. {amount} withdrawn from {self.owner}'s account.")
        else:
            print(f"Insufficient balance in {self.owner}'s account.")
        print(f"Remaining balance for {self.owner}'s account is Rs. {self.balance} ")
        print()

    def display_statement(self):
        print(f"Account balance for {self.owner}: Rs. {self.balance}")
        print()

acc = BankAccount("Aashish Awasthi")
acc.deposit(500)
acc.deposit(3000)
acc.withdraw(1000)
acc.display_statement()
acc.withdraw(6000)
acc.deposit(-400)
acc.withdraw(0)
```

***2. The Digital Shopping Cart***

**Goal:** Manage an internal data structure (a dictionary) where tracking totals requires calling distinct helper operations.
- **Class Name:** ShoppingCart
- **The Methods:**
  - __init__(self): Starts with an empty dictionary self.items where keys are item names and values are prices.
  - add_item(self, name, price): Adds a product name and its price to the dictionary.
  - calculate_subtotal(self): Iterates through the dictionary to sum and return the total price of all items.
  - apply_discount(self, discount_percentage): Calls calculate_subtotal to find the price, figures out the discount amount, and returns the final checkout price.
- **How They Work Together:** apply_discount cannot function on its own; it forces Python to execute calculate_subtotal first to grab the baseline data.

```python
class ShoppingCart:
    def __init__(self):
        self.items = {}

    def add_item(self, name, price):
        if name not in self.items:
            self.items[name] = price
            print(f"{name} added to the cart with price Rs. {price}")

    def calculate_subtotal(self):
        subtotal = 0
        for price in self.items.values():
            subtotal += price
        return subtotal

    def apply_discount(self, discount_percentage):
        subtotal = self.calculate_subtotal()
        discount_amount = subtotal * (discount_percentage / 100)
        final_price = subtotal - discount_amount
        return final_price

shopping = ShoppingCart()
shopping.add_item("TV", 25000)
shopping.add_item("Dryer", 1200)
print(f"Subtotal: Rs. {shopping.calculate_subtotal()}")
print(f"Final Price with 10% discount: Rs. {shopping.apply_discount(10)}")
```

**Exercise 3:**

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

rect = Rectangle(10, 4)
print(f"Area = {rect.area()} and Perimeter = {rect.perimeter()}")
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)

s1 = Student("Alice", [85, 90, 78, 92, 88])
print(f"{s1.name}'s Average Grade: {s1.average()}")
```

```python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_value(self):
        return self.price * self.quantity

p1 = Product("Laptop", 899.99, 5)
print(f"Total stock value of {p1.name}: ${p1.total_value():.2f}")
```

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Balance after deposit: {self.balance}")
        else:
            print("Invalid deposit amount. Amount must be greater than zero.")

    def withdrawal(self, amount):
        if amount < 0:
            print("Invalid withdrawal amount. Amount must be greater than zero.")
        else:
            if amount <= self.balance:
                self.balance -= amount
                print(f"Balance after withdrawal: {self.balance}")
            else:
                print(f"Insufficient funds. Current balance: {self.balance}")

bank = BankAccount(1000)
bank.deposit(500)
bank.withdrawal(200)
bank.withdrawal(2000)
```
