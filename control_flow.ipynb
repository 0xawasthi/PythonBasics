# Control Flow in Python

## Conditional Statement in Python (if, elif, else)
- if statement
- else statement
- elif statement
- nested
- few examples

```python
# if statement
## if <condition>:
      #<statement>

age = 21
if age >= 18:
  print("budo vayecha")
```

```python
# else statement

age = 16
if age >= 18:
    print("budo vayecha")
else:
    print("baccha raicha")
```

```python
# elif statement

num = int(input("Enter a number:"))
if num > 0:
  if num % 2 == 0:
    print("Even number")
  else:
    print("Odd number")
elif num < 0:
  print("Negative number")
else:
  print("Zero")
```

```python
# Determine if a year is a leap year or not using a nested condition statement

year = int(input("Enter a year:"))
if year % 4 == 0:
    print("It is a leap year.")
else:
    print("It is not a leap year.")
```

```python
# Ticket price determination

age = int(input("Enter your age:"))
if age < 5:
    print("Ticket is free.")
elif age >= 5 and age <13:
    print("Ticket price is $10.")
elif age >= 13 and age < 18:
    student = input("Are you a student? (yes/no):")
    if student == "yes":
        print("Ticket price is $12.")
    else:
        print("Ticket price is $15.")
elif age >= 18 and age < 65:
    student = input("Are you a student? (yes/no):")
    if student == "yes":
        print("Ticket price is $18.")
    else:
        print("Ticket price is $25.")
else:
    print("Ticket price is $20.")
```

```python
# Employee Bonus Calculation

year = int(input("Enter Years of Service:"))
rating = float(input("Enter Performance Rating (1.0 - 5.0):"))
salary = int(input("Enter Salary:"))

if rating >= 4.5 and rating <= 5.0:
    if year >= 10:
        bonus = salary * 0.20
    elif year >=5:
        bonus = salary * 0.15
    else:
        bonus = salary * 0.10
elif rating >= 3.5 and rating < 4.5:
    if year >= 10:
        bonus = salary * 0.15
    elif year >=5:
        bonus = salary * 0.10
    else:
        bonus = salary * 0.05
elif rating <3.5:
    bonus = 0
else:
    print("Invalid Performance Rating")

print(f"Bonus: {bonus:.2f}")
```

# Loops in Python
- For Loop
- While Loop
- Nested Loop
- Examples

```python
# Range Function
# range(start, stop, step)

# for i in range(5):
#     print(i)


for i in range(1, 10, 2):
    print(i)
```

```python
# string

str = "Hello World"
for i in str:
    print(i)
```

```python
# While loop

count = 0

while count<5:
    print(count)
    count = count + 1
```

Loop Control Statement (break, continue, pass)

```python
# Calculate the sum of first n natural numbers using for loop

n = int(input("Enter a number:"))
sum = 0
for i in range(1, n+1):
    sum = sum + i
print(sum)
```

```python
# Calculate the sum of first n natural numbers using for loop

n = int(input("Enter a number:"))
sum = 0
i = 1
while i <= n:
    sum = sum + i
    i = i + 1
print(sum)
```

```python
# Prime number between 1 and 100 using for loop

for i in range(1, 101):
    for j in range(2, i):
        if i % j == 0:
            break
    else:
        print(i)
```
