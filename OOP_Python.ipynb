### Multiple Methods

A class can have multiple methods that work together.

````python
class Playlist:

    def __init__(self, title):
        self.title = title
        self.songs = []

    def add_song(self, song):
        self.songs.append(song)
        print(f"Added a song: {song}")

    def remove_song(self, song):
        if song in self.songs:
            self.songs.remove(song)
            print(f"Removed a song: {song}")

    def show_song(self):
        print(f"Playlist: {self.title}")
        for song in self.songs:
            print(f"- {song}")

my_playlist = Playlist("My Playlist")
my_playlist.add_song("Song 1")
my_playlist.add_song("Song 2")
my_playlist.add_song("Song 3")
my_playlist.show_song()
````

## Inheritance in Python

Inheritance allows us to define a class that inherits all the methods and properties from another class
- **Parent class**: Class being inherited from, also called it as **base class**.
- **Child class**: Class that inherits from another class, also called it as **derived class**.

````python
#@ Create a parent class
# Any class can be a parent class, so the syntax remain same as above class defination

class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def printname(self):
        print(self.firstname, self.lastname)

# Use the Person class to create an object, and then execute the printname method:

person1 = Person("Aashish", "Awasthi")
person1.printname()
````

````python
#@ Create a child class
'''To create a class that inherits the functionality from another class, send the
 parent class as a parameter when creating the child class...'''

class Student(Person):
    pass
````

````python
person2 = Student("Rebanta", "Aryal")
person2.printname()
````

**Multiple Inheritance:**

image.png

````python
class Mother: #Parent class 1
    mothername = ""

    def mother(self):
        print(self.mothername)

class Father: # Parent class 2
    fathername = ""

    def father(self):
        print(self.fathername)

class Son(Mother, Father): # Child Class

    def parents(self):
        print("Father :", self.fathername)
        print("Mother :", self.mothername)

s1 = Son()
s1.fathername = "Ram"
s1.mothername = "Sita"
s1.parents()
````

**Multilevel Inheritance:**

image.png

````python
class GrandFather:
    def __init__(self, grandfathername):
        self.grandfathername = grandfathername

class Father(GrandFather):
    def __init__(self, fathername, grandfathername):
        self.fathername = fathername

        # invoking constructor of Grandfather class
        GrandFather.__init__(self, grandfathername)

class Son(Father):
    def __init__(self, sonname, fathername, grandfathername):
        self.sonname = sonname

        # invoking constructor of Father class
        Father.__init__(self, fathername, grandfathername)

    def print_name(self):
        print('Grandfather name :', self.grandfathername)
        print("Father name :", self.fathername)
        print("Son name :", self.sonname)

s1 = Son('Prince', 'Rampal', 'Lalmani')
# print(s1.grandfathername)
s1.print_name()
````

**Important Notes:**
- Son inherits from Father and Father inherits from Grandfather
- Each constructor passes values up the inheritance chain using explicit constructor calls
- All ancestor class attributes are accessible from the bottom most class which is Son is our case

### Super() Function
 - super() function that will make the child class inherit all the methods and properties from its parent class

 - It is commonly used in the child class's __init__() method to initialize inherited attributes.

````python
#@ Animal
# Parent Class

class Animal:
    def __init__(self, name):
        self.name = name

    def info(self):
        print("Animal Name: ", self.name)

class Dog(Animal):
    def __init__(self, name, breed):
      # Call constructor based upon python's method resolution order (MRO)
        super().__init__(name)
        self.breed = breed

    def details(self):
        print(self.name, "is a", self.breed)

dog1 = Dog("Khaire", "Bhusiya")
dog1.info()
dog1.details()
````

***Takeaway:***

- ```super()``` function is used inside ```__init__()``` method of Dog to call the constructor of ```Animal``` and initialize inherited atttribute (name)
- It ensures that parent class functionality is reused without needing to rewrite the code in the child class.

- **Method Overloading:**

Python,  the real comiple time polymorphism (overloading) doesnt support. why?
It is beacuse, the same behaviour can be achieved by using variable length argument. (*args, **kwargs)

````python
#@ Example

class Calculator:
    def multiply(self, a=1, b=1, *args):
        result = a * b
        for num in args:
            result *= num

        return result

calc = Calculator()
print(calc.multiply())
print(calc.multiply(7))
print(calc.multiply(10, 5))
print(calc.multiply(10, 5, 2))
print(calc.multiply(1, 2, 3, 4, 5))
````

- **Method Overriding:**

````python
#@ overriding example

class Animal:
    def make_sound(self):
        print("Animal is making a sound")

class Dog(Animal):
    def make_sound(self):
        return "Woof Woof!!"

class Cat(Animal):
    def make_sound(self):
        return "Meow Meow!"

animals = [Dog(), Cat(), Animal()]
for animal in animals:
    print(animal.make_sound())
````

Sound method behaves differently depending on whether the object is Dog or Cat or Animal, and this type of decision is made during the runtime. Thus, this dynamic nature makes Python particularly powerful for runtime Polymorphism,

**Practice Question:** *Polymorphism with Functions and Methods*

Create a base class Shape with a method area().

Then create two derived classes:

- Rectangle — takes width and height and calculates the area.
- Circle — takes radius and calculates the area.

Finally, create a function print_area(shape) that accepts any shape object and prints its area.

````python
class Shape:

    def area(self):
        pass

class Rectangle(Shape):

    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

def print_area(shape):
    print(shape.area())

rectangle = Rectangle(10, 20)
circle = Circle(5)

print_area(rectangle)
print_area(circle)
````
