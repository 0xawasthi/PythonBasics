```python
# Create a Python To-Do List with three tasks. Add two new tasks, remove one completed task, check if "Pay bills" exists, and print the remaining tasks.

to_do_list = ["Shopping", "Pay Bills", "Drink Water"]

to_do_list.append("Call mom")
to_do_list.append("Clean house")

to_do_list.remove("Shopping")

if "Pay Bills" in to_do_list:
    print("Pay Bills exists.")
else:
    print("Pay Bills does not exist.")

print("\nRemaining tasks:")
for task in to_do_list:
    print(f" - {task}")
```

```python
# Organizing Student Grades Create a list to store and calculate average grades for students

student_grades = [45, 80, 66, 72, 77, 69]

average_grade = sum(student_grades) / len(student_grades)

print(f"Average grade: {average_grade:.2f}")
```

Count Even and Odd Numbers

**Given Input:** Numbers: [10, 21, 4, 45, 66, 93, 11]

**Expected Output:**

Even numbers: 3

Odd numbers: 4


```python
numbers = [10, 21, 4, 45, 66, 93, 11]

print(f"Even numbers: {len([num for num in numbers if num % 2 == 0])}")
print(f"Odd numbers: {len([num for num in numbers if num % 2 != 0])}")
```
